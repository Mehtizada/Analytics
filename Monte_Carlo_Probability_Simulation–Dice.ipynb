```python
# MONTE CARLO – DICE ROLL SIMULATION

import random
import matplotlib.pyplot as plt
```

```python
# 10,000 dice rolls

results = [random.randint(1, 6) for _ in range(10000)]
```

```python
plt.hist(results, bins=6, edgecolor='black')
plt.title("Monte Carlo Dice Roll Simulation")
plt.xlabel("Dice Result")
plt.ylabel("Frequency")
plt.show()
```

```python
probabilities = {i: results.count(i)/len(results) for i in range(1,7)}
print("Probabilities:", probabilities)
```
